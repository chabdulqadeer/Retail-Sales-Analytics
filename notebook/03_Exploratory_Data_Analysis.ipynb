```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Importing the dataset
df = pd.read_csv(r"F:\Retail-Sales-Analytics\data\clean\cleaned_Superstore.csv", encoding='latin1')
```

```python
# KPI 1
Total_Sales = round(df['Sales'].sum(),2)
print(f'Total Revenue:${Total_Sales:,.2f}')
```

```python
#KPI 2
Total_Profit = round(df['Profit'].sum(),2)
print(f'Total Profit:${Total_Profit:,.2f}')
```

```python
# KPI 3
profit_margin = (Total_Profit * 100)/Total_Sales
print(f'The Profit margin: {round(profit_margin,2)}%')
```

```python
# KPI 4
total_order = df['Order ID'].nunique()
print(f"Total number of Orders: {total_order}")
```

```python
# KPI 5
customers = df['Customer ID'].nunique()
print(f"No.of Customers: {customers}")
```

```python
# KPI 6
average_order_value = Total_Sales/total_order
print(f"Average Order Value : ${round(average_order_value,2):,.2f}")
```

```python
# KPI 7
average_profit_order = Total_Profit/total_order
print(f"Average Profit per order:${round(average_profit_order,2):,.2f}")
```

```python
# KPI 8
top_sales_products = df.groupby(df['Product Name'])['Sales'].sum().sort_values(ascending=False).head(10)
top_sales_products
```

```python
# KPI 9
top_profit_product = df.groupby(df['Product Name'])['Profit'].sum().sort_values(ascending=False).head(10)
top_profit_product
```

```python
# KPI 10
top_customers = df.groupby(df['Customer Name'])['Sales'].sum().sort_values(ascending=False).head(10)
top_customers
```

```python
# KPI 11
top_customers_profit = df.groupby(df['Customer Name'])['Profit'].sum().sort_values(ascending=False).head(10)
top_customers_profit
```

```python
# Regional Analysis
# KPI 12
regional_product = df.groupby(['Region'])['Sales'].sum().sort_values(ascending=False).head(10)
regional_product
```

```python
# KPI 13
regional_product_profit = df.groupby(['Region'])['Profit'].sum().sort_values(ascending=False).head(10)
regional_product_profit
```

```python
# KPI 14
top_category =  df.groupby('Category')['Sales'].sum().sort_values(ascending=False).head(10)
top_category
```

```python
# KPI 15
top_category_profit =  df.groupby('Category')['Profit'].sum().sort_values(ascending=False).head(10)
top_category_profit
```

```python
data =  pd.pivot_table(
    data=df,
    index='Region',
    values=['Sales','Profit','Customer Name','Order ID'],
    aggfunc={'Sales':'sum','Profit':'sum','Customer Name':'nunique','Order ID':'nunique'}
)
data['Profit Margin (%)'] = (data['Profit'] / data['Sales']) * 100
data['Order per Customers'] = data['Order ID']/data['Customer Name']
data = data.rename(columns={
    'Customer Name': 'No.of Customers',
    'Order ID': 'No.of Orders'
})
data.round(2)
```
