```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r'F:\Retail-Sales-Analytics\data\clean\cleaned_Superstore.csv', encoding='latin1')
```

```python
# Visualization 1 — Sales by Category
category_sales = df.groupby('Category')['Sales'].sum().sort_values()
plt.bar(
    x=category_sales.index,
    height=category_sales.values
)
plt.title('Sales by Category')
plt.xlabel('Categories')
plt.ylabel('Sales')
plt.show()
```

```python
# Visualization 1 — Sales by Category
category_sales = df.groupby('Category')['Sales'].sum().sort_values()
category_sales.plot(kind='bar', figsize=(8,5))
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Visualization 2 — Profit by Category
category_profit = df.groupby('Category')['Profit'].sum().sort_values()
category_profit.plot(kind='barh')
plt.title("Profit by Category")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Visualization 3 — Sales by Region
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
plt.bar(x=region_sales.index, height=region_sales.values)
plt.title('Region VS Sales')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.show()
```

```python
# Visualization 3 — Profit by Region
region_sales = df.groupby('Region')['Profit'].sum().sort_values(ascending=False)
plt.bar(x=region_sales.index, height=region_sales.values)
plt.title('Region VS Profit')
plt.xlabel('Region')
plt.ylabel('Profit')
plt.show()
```

```python
# Visualization 5 — Top 10 Products by Sales
top_product = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)
short_names = top_product.index.str.slice(0, 35)
plt.barh(
    y=short_names,
    width=top_product.values,
    height=0.8
)
plt.title('Top 10 Products by Sales')
plt.Figure(figsize=(14,7))
plt.xlabel('Sales')
plt.ylabel('Products')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Visualization 6 — Top 10 Products by Profit
top_product_profit = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)
short_names = top_product_profit.index.str.slice(0, 35)
plt.barh(
    y=short_names,
    width=top_product_profit.values,
    height=0.8
)
plt.title('Top 10 Products by Profit')
plt.xlabel('Profit')
plt.ylabel('Products')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Bonus Analysis — Discount vs Profit
discount_profit = df.groupby('Discount')['Profit'].mean().sort_index()
plt.plot(
    discount_profit.index,
    discount_profit.values,
    marker='o'
)
plt.title('Discount vs Profit')
plt.xlabel('Discount')
plt.ylabel('Averge Profit')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
